```python
import pandas as pd
import numpy as np
import sys
import os
import matplotlib.pyplot as plt

# Define the path to the src directory manually
src_path = os.path.abspath('../src')
sys.path.insert(0, src_path)

import module_cost_calculations

# Use color-blind friendly style
plt.style.use('tableau-colorblind10')

unit_file_path = '../data/SitkanaCosts_Feb2025_unit.csv'
system_file_path = '../data/SitkanaCosts_Feb2025_system.csv'

system_df = pd.read_csv(system_file_path) 
unit_df = pd.read_csv(unit_file_path) 
```

# Check Unit Cost

```python
turbine_radius_m = np.linspace(np.min(unit_df['Rotor Radius (m)']),np.max(unit_df['Rotor Radius (m)']),50)
turbine_rated_power_W = np.linspace(np.min(unit_df['Power Rating (W)']),np.max(unit_df['Power Rating (W)']),50)
number_of_turbines=1
electrical_cable_length_m=None
mooring_cable_length_m=None
force_vessel_drag_N=None
force_turbine_thrust_N=None
vessel_volume_m3=None
BatteryCapacity_kWh=None
```

```python
# Calculate costs
rotor_cost = module_cost_calculations.calculate_rotor_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)

rotor_construction_cost = module_cost_calculations.calculate_rotor_construction_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)

steel_component_cost = module_cost_calculations.calculate_steel_component_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)


# Create subplots
fig, axs = plt.subplots(3, 1, figsize=(10, 8))  # Set figure size

# Plot Rotor Cost
axs[0].plot(unit_df['Rotor Radius (m)'], unit_df['Rotor Cost ($)'], '.', label='Customer Cost')
axs[0].plot(turbine_radius_m, rotor_cost, '-', label='Calculated Cost')
# axs[0].set_xlabel('Rotor Radius (m)')
axs[0].set_ylabel('Rotor Cost ($)')
axs[0].legend()
axs[0].grid(True)

# Plot Rotor Construction Cost
axs[1].plot(unit_df['Rotor Radius (m)'], unit_df['Rotor Construction Cost ($)'], '.', label='Customer Cost')
axs[1].plot(turbine_radius_m, rotor_construction_cost, '-', label='Calculated Cost')
# axs[1].set_xlabel('Rotor Radius (m)')
axs[1].set_ylabel('Rotor Construction Cost ($)')
axs[1].legend()
axs[1].grid(True)

axs[2].plot(unit_df['Rotor Radius (m)'], unit_df['Steel Components Cost ($)'], '.', label='Customer Cost')
axs[2].plot(turbine_radius_m, steel_component_cost, '-', label='Calculated Cost')
axs[2].set_xlabel('Rotor Radius (m)')
axs[2].set_ylabel('Steel Component Cost ($)')
axs[2].legend()
axs[2].grid(True) 


plt.tight_layout()

plt.savefig('rotor_costs.png', format='png',dpi=300)
 
plt.show()
```

```python

generator_cost = module_cost_calculations.calculate_generator_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)

assembly_cost = module_cost_calculations.calculate_assembly_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)

concrete_cost = module_cost_calculations.calculate_concrete_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)

gearbox_cost = module_cost_calculations.calculate_gearbox_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)

charge_controller_cost = module_cost_calculations.calculate_charge_controller_cost_SITKANA(
    turbine_radius_m, 
    turbine_rated_power_W, 
    number_of_turbines, 
    electrical_cable_length_m, 
    mooring_cable_length_m, 
    force_vessel_drag_N, 
    force_turbine_thrust_N, 
    vessel_volume_m3=None, 
    BatteryCapacity_kWh=None
)

# Create subplots
fig, axs = plt.subplots(5, 1, figsize=(10, 8/3*5))  # Set figure size

axs[0].plot(unit_df['Power Rating (W)'], unit_df['Generator Cost ($)'], '.', label='Customer Cost')
axs[0].plot(turbine_rated_power_W, generator_cost, '-', label='Calculated Cost')
# axs[0].set_xlabel('Power Rating (W)')
axs[0].set_ylabel('Generator Cost ($)')
axs[0].legend()
axs[0].grid(True)

axs[1].plot(unit_df['Power Rating (W)'], unit_df['Assembly Cost ($)'], '.', label='Customer Cost')
axs[1].plot(turbine_rated_power_W, assembly_cost, '-', label='Calculated Cost')
# axs[1].set_xlabel('Power Rating (W)')
axs[1].set_ylabel('Assembly Cost ($)')
axs[1].legend()
axs[1].grid(True)

axs[2].plot(unit_df['Power Rating (W)'], unit_df['Concrete Cost ($)'], '.', label='Customer Cost')
axs[2].plot(turbine_rated_power_W, concrete_cost, '-', label='Calculated Cost')
# axs[2].set_xlabel('Power Rating (W)')
axs[2].set_ylabel('Concrete Cost ($)')
axs[2].legend()
axs[2].grid(True)

axs[3].plot(unit_df['Power Rating (W)'], unit_df['Gearbox Cost ($)'], '.', label='Customer Cost')
axs[3].plot(turbine_rated_power_W, gearbox_cost, '-', label='Calculated Cost')
# axs[3].set_xlabel('Power Rating (W)')
axs[3].set_ylabel('Gearbox Cost ($)')
axs[3].legend()
axs[3].grid(True)

axs[4].plot(unit_df['Power Rating (W)'], unit_df['Charge Controller Cost ($)'], '.', label='Customer Cost')
axs[4].plot(turbine_rated_power_W, charge_controller_cost, '-', label='Calculated Cost')
axs[4].set_xlabel('Power Rating (W)')
axs[4].set_ylabel('Charge Controller Cost ($)')
axs[4].legend()
axs[4].grid(True) 

plt.tight_layout()

plt.savefig('prated_costs.png', format='png',dpi=300)

# Adjust layout
plt.show()
```

# System Level Cost

```python

number_of_turbines = 2.0
anchor_cost = module_cost_calculations.calculate_anchor_cost_SITKANA(turbine_radius_m, 
                            turbine_rated_power_W, 
                            number_of_turbines, 
                            electrical_cable_length_m, 
                            mooring_cable_length_m, 
                            force_vessel_drag_N, 
                            force_turbine_thrust_N, 
                            vessel_volume_m3=None, 
                            BatteryCapacity_kWh=None)

oprating_cost = module_cost_calculations.operating_cost_SITKANA(turbine_rated_power_W, number_of_turbines)

fig, axs = plt.subplots(2, 1, figsize=(10, 8/3*2))  # Set figure size

axs[0].plot(system_df['System Size (W)'], system_df['Anchor Cost ($)'], '.', label='Customer Cost')
axs[0].plot(turbine_rated_power_W*number_of_turbines, anchor_cost, '-', label='Calculated Cost')
# axs[0].set_xlabel('Power Rating (W)')
axs[0].set_ylabel('Anchor Cost ($)')
axs[0].legend()
axs[0].grid(True)

axs[1].plot(system_df['System Size (W)'], system_df['Operation Cost ($)'], '.', label='Customer Cost')
axs[1].plot(turbine_rated_power_W*number_of_turbines, oprating_cost, '-', label='Calculated Cost')
# axs[0].set_xlabel('Power Rating (W)')
axs[1].set_ylabel('Operating Cost ($)')
axs[1].legend()
axs[1].grid(True)
plt.tight_layout()

plt.savefig('system_costs.png', format='png',dpi=300)

# Adjust layout
plt.show()
```
