```python
import sys
import os

# Set the current and parent paths
current_path = os.getcwd()
parent_path = os.path.dirname(current_path)

# Add the parent path to the system path
sys.path.append(parent_path)

# Import the constVessel module for creating vessel properties
import constVessel
```

## Creating a New Vessel

The `constVessel` module contains the function `create_vessel`, which can be used to create a new vessel.

### Function Inputs

The inputs to the `create_vessel` function must follow the order below:

1. **Name**: Name of the vessel.
2. **Xm**: Horizontal distance from the center of gravity (CG) to the mooring line [meters].
3. **Zm**: Vertical distance from the center of gravity (CG) to the mooring line [meters].
4. **Km**: Pitch stiffness [N*m/rad].
5. **Theta**: Mooring line angle [degrees].
6. **Phi**: Vessel pitch constraint [degrees].
7. **Area**: Area of the vessel [square meters].
8. **Cd**: Vessel drag coefficient.

```python
# Define vessel properties
Name = "MyShip"       # Name of the vessel
Xm = 10.0             # Horizontal distance from CG to mooring line [meters]
Zm = 1.0              # Vertical distance from CG to mooring line [meters]
Km = 100.0e6          # Pitch stiffness [N*m/rad]
theta = 45.0          # Mooring line angle [degrees]
phi = 10.0            # Vessel pitch constraint [degrees]
area = 15.0           # Area of the vessel [square meters]
Cd = 1.0              # Vessel drag coefficient

# Create a new vessel using the defined properties
NewBoat = constVessel.create_vessel(Name, Xm, Zm, Km, theta, phi, area, Cd)
```

## Verifying Vessel Creation

To ensure that the vessel has been created correctly, we can print its details using the `print_vessel_details` function from the `constVessel` module.


```python
# Print the details of the newly created vessel
constVessel.print_vessel_details(NewBoat)
```
